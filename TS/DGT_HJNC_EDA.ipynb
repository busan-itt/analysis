HJNC 기준 반입 S / 반출 T -> 수입환적(TI) = 20,179건

```python
# 1. 비교할 컬럼 설정
col_in = '반입차량번호'
col_hjnc_out = 'HJNC_반출차량번호'

col_out = '반출차량번호'
col_hjnc_in = 'HJNC_반입차량번호'

# 2. 두 컬럼 모두 값이 존재하는(결측치가 아닌) 행들만 대상으로 일치 여부 확인
# (어느 한쪽이라도 비어있으면 비교 대상에서 제외하거나 별도 카운트)

# 조건 1: 반입차량번호와 HJNC_반출차량번호 비교 (둘 다 값이 있는 경우)
both_exist_1 = df[col_in].notna() & df[col_hjnc_out].notna()
match_1 = (df.loc[both_exist_1, col_in] == df.loc[both_exist_1, col_hjnc_out])

# 조건 2: 반출차량번호와 HJNC_반입차량번호 비교 (둘 다 값이 있는 경우)
both_exist_2 = df[col_out].notna() & df[col_hjnc_in].notna()
match_2 = (df.loc[both_exist_2, col_out] == df.loc[both_exist_2, col_hjnc_in])

# 3. 결과 요약 출력
print("--- [HJNC->DGT: 반입차량번호 & HJNC_반출차량번호] ---")
print(f"둘 다 값이 존재하는 행 수: {both_exist_1.sum()}개")
print(f"그 중 값이 일치하는 행 수: {match_1.sum()}개")
print(f"값이 불일치하는 행 수: {(~match_1).sum()}개\n")

print("--- [DGT->HJNC: 반출차량번호 & HJNC_반입차량번호] ---")
print(f"둘 다 값이 존재하는 행 수: {both_exist_2.sum()}개")
print(f"그 중 값이 일치하는 행 수: {match_2.sum()}개")
print(f"값이 불일치하는 행 수: {(~match_2).sum()}개")

# 4. 값이 둘 다 존재하는데 불일치하는 행 => 별도 확인 필요
diff_rows_1 = df[both_exist_1 & (df[col_in] != df[col_hjnc_out])]
display(diff_rows_1[['컨테이너번호', '반입시간', '반출시간', col_in, col_hjnc_out]])

diff_rows_2 = df[both_exist_2 & (df[col_out] != df[col_hjnc_in])]
display(diff_rows_2[['컨테이너번호', '반입시간', '반출시간', col_out, col_hjnc_in]])
```

총 19건에 대하여 화물관리번호가 일치하며, 유니패스 조회하여 HJNC <-> DGT 이동 이력 확인하였음.
모종의 이유로 터미널 간 이동 시 차량이 변경되었거나, 차량번호 기입 오류 등으로 예상됨.
