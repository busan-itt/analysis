## 1. DGT -> HJNC 타부두 환적 건 분석

```python
# DGT -> HJNC 타부두 환적 건만 추출
DGT_to_HJNC_df = filtered_2_df[filtered_2_df['Class'] == 'TI'].copy()
DGT_to_HJNC_df
```

```python
# 계산을 위해 반입시간, 반출시간, 온샤시 시간, 야드 장치 시간 자료 형식 날짜로 변경
DGT_to_HJNC_df['반입시간'] = pd.to_datetime(DGT_to_HJNC_df['반입시간'])
DGT_to_HJNC_df['반출시간'] = pd.to_datetime(DGT_to_HJNC_df['반출시간'])
DGT_to_HJNC_df['온샤시 시간'] = pd.to_datetime(DGT_to_HJNC_df['온샤시 시간'])
DGT_to_HJNC_df['야드 장치 시간'] = pd.to_datetime(DGT_to_HJNC_df['야드 장치 시간'])
DGT_to_HJNC_df['HJNC_반입시간'] = pd.to_datetime(DGT_to_HJNC_df['HJNC_반입시간'])
DGT_to_HJNC_df['HJNC_반출시간'] = pd.to_datetime(DGT_to_HJNC_df['HJNC_반출시간'])
DGT_to_HJNC_df['HJNC_온샤시 시간'] = pd.to_datetime(DGT_to_HJNC_df['HJNC_온샤시 시간'])
DGT_to_HJNC_df['HJNC_야드 장치 시간'] = pd.to_datetime(DGT_to_HJNC_df['HJNC_야드 장치 시간'])
```

```python
DGT_to_HJNC_df.info()
```

```python
# 총 소요시간 구하기
DGT_to_HJNC_df['총소요시간'] = (DGT_to_HJNC_df['반출시간'] - DGT_to_HJNC_df['반입시간']).dt.total_seconds() / 3600  # 시간 단위
# 만약 '분(minute)' 단위로 구하고 싶다면 아래 주석처럼 60으로 나누거나 .dt.total_seconds() / 60을 사용하세요
# df['총소요시간'] = (df['반출시간'] - df['반입시간']).dt.total_seconds() / 60  # 분 단위

DGT_to_HJNC_df['HJNC_총소요시간'] = (DGT_to_HJNC_df['HJNC_반출시간'] - DGT_to_HJNC_df['HJNC_반입시간']).dt.total_seconds() / 3600  # 시간 단위
DGT_to_HJNC_df
```

```python
# 총 소요시간 구하기
DGT_to_HJNC_df['환적총시간(터미널간이동시간제외)'] = (DGT_to_HJNC_df['총소요시간'] + DGT_to_HJNC_df['HJNC_총소요시간'])
DGT_to_HJNC_df['환적총시간'] = (DGT_to_HJNC_df['HJNC_반출시간'] - DGT_to_HJNC_df['반입시간']).dt.total_seconds() / 3600  # 시간 단위

DGT_to_HJNC_df
```

```python
mean_val_1 = DGT_to_HJNC_df['환적총시간(터미널간이동시간제외)'].mean()
median_val_1 = DGT_to_HJNC_df['환적총시간(터미널간이동시간제외)'].median()
mean_val_2 = DGT_to_HJNC_df['환적총시간'].mean()
median_val_2 = DGT_to_HJNC_df['환적총시간'].median()

print(f'환적총시간(터미널이동시간제외) 평균 : {mean_val_1:.2f}시간')
print(f'환적총시간(터미널이동시간제외) 중앙값 : {median_val_1:.2f}시간')
print(f'환적총시간 평균 : {mean_val_2:.2f}시간')
print(f'환적총시간 중앙값 : {median_val_2:.2f}시간')
```

## 각 공정별 작업 시간 계산

자부두 환적과는 달리, 타부두 환적 시에는 터미널 이동 시간 및 두 번째 부두에서의 작업 과정이 추가됨.
- 자부부 환적 : 반입 -> 야드장치 -> 온샤시 -> 반출 (공정 A, B, C)
- 타부두 환적 : 반입 -> 야드장치 -> 온샤시 -> 반출 -> 터미널 이동 -> 반입 -> 야드장치 -> 온샤시 -> 반출 (공정 A, B, C, 이동, C', B', A')
